```python
#The dataset includes 97 rows and 6 columns(before cleaning) with columns that consist of date, item, city, price.
#The problems that could be diagnosed by me include that of missing data, duplication and outlier value.
#Below is the cleaning code suggested buy AI.
import pandas as pd
import numpy as np

# Load original dataset and create a working copy
df = pd.read_csv('teaching_prices_messy.csv')
df_clean = df.copy()

# --- STEP 2: TARGETED CLEANING CODE ---

# 1. Remove commas from 'price' so numbers like "1,250" don't become NaN, then convert to numeric
df_clean['price'] = df_clean['price'].astype(str).str.replace(',', '', regex=False)
df_clean['price'] = pd.to_numeric(df_clean['price'], errors='coerce')

# 2. Flag unusually large prices (e.g., > 500) for review instead of deleting them
df_clean['price_flag'] = np.where(df_clean['price'] > 500, 'outlier_check', 'ok')

# 3. Remove exact duplicate rows
df_clean = df_clean.drop_duplicates()

# 4. Keep the 1 missing price row as NaN (do not guess or drop the rest of the record)
# (No deletion needed; documented in cleaning log)

# --- STEP 3: BEFORE-AND-AFTER VERIFICATION ---

comparison = pd.DataFrame({
    'Before (Raw)': [len(df), df.isna().sum().sum(), df.duplicated().sum()],
    'After (Clean)': [len(df_clean), df_clean.drop(columns=['price_flag']).isna().sum().sum(), df_clean.duplicated().sum()]
}, index=['Total Rows', 'Total Missing Values', 'Exact Duplicates'])

print("=== BEFORE VS AFTER SUMMARY ===")
display(comparison)

print("\n=== MISSING VALUES PER COLUMN ===")
display(pd.DataFrame({
    'Before': df.isna().sum(),
    'After': df_clean.drop(columns=['price_flag']).isna().sum()
}))

print("\n=== PRICE SUMMARY STATS (AFTER CLEANING) ===")
display(df_clean.describe())

print("\n=== ITEM FREQUENCIES (CHECK NO VALID ITEMS LOST) ===")
print(df_clean['item'].value_counts(dropna=False))

print("\n=== FLAGGED OUTLIER ROW(S) ===")
display(df_clean[df_clean['price_flag'] == 'outlier_check'])
```
